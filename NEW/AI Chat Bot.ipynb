```python
from llama_cpp import Llama

# Load the AI model
llm = Llama.from_pretrained(
    repo_id="TheBloke/zephyr-7B-beta-GGUF",
    filename="zephyr-7b-beta.Q4_K_M.gguf",
    n_ctx=4096,
    n_gpu_layers=0
)

# Store chat history
history = [
    {
        "role": "system",
        "content": "You are a helpful AI assistant."
    }
]

print("================================")
print("        AI CHAT BOT")
print("================================")
print("Type 'exit' to stop the chatbot.\n")


while True:

    # Get user input
    prompt = input("You: ")

    # Exit chatbot
    if prompt.lower() in ["exit", "quit", "bye"]:
        print("Exiting chat.")
        break

    # Add user message
    history.append(
        {
            "role": "user",
            "content": prompt
        }
    )

    # Generate response
    output = llm.create_chat_completion(
        messages=history
    )

    # Get assistant response
    reply = output["choices"][0]["message"]["content"]

    # Store response
    history.append(
        {
            "role": "assistant",
            "content": reply
        }
    )

    # Display response
    print("Assistant:", reply)
    print()
```
