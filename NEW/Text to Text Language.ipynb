```python
!pip install transformers sentencepiece torch

from transformers import M2M100ForConditionalGeneration
from transformers import M2M100Tokenizer

model_name = "facebook/m2m100_418M"

# Load tokenizer and model
tokenizer = M2M100Tokenizer.from_pretrained(model_name)
model = M2M100ForConditionalGeneration.from_pretrained(model_name)

languages = {
    "1": ("English", "en"),
    "2": ("Hindi", "hi"),
    "3": ("Marathi", "mr"),
    "4": ("Gujarati", "gu"),
    "5": ("French", "fr"),
    "6": ("German", "de"),
    "7": ("Spanish", "es"),
    "8": ("Japanese", "ja"),
    "9": ("Arabic", "ar")
}

print("Languages:")

for number, (name, code) in languages.items():
    print(number, name)

source = input("\nSource language number: ")
destination = input("Destination language number: ")
text = input("Enter text: ")

source_name, source_code = languages[source]
destination_name, destination_code = languages[destination]

# Set source language
tokenizer.src_lang = source_code

# Convert text into tokens
encoded = tokenizer(text, return_tensors="pt")

# Generate translation
output = model.generate(
    **encoded,
    forced_bos_token_id=tokenizer.get_lang_id(destination_code)
)

# Convert tokens to text
translation = tokenizer.decode(
    output[0],
    skip_special_tokens=True
)

print("\nOriginal:", text)
print("Translation:", translation)
```
