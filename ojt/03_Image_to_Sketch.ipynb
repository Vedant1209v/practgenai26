```python
from pathlib import Path
import cv2
import matplotlib.pyplot as plt

image_path = r"C:\Users\landg\OneDrive - PTVA's Mulund College of Commerce\Desktop\Nikhil.jpg"

if not Path(image_path).is_file():
    raise FileNotFoundError(f"Image not found: {image_path}")

img = cv2.imread(image_path)

if img is None:
    raise ValueError("OpenCV could not read this image. Try a JPG or PNG file.")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

sketch = cv2.Canny(gray, 50, 150)

out_path = str(
    Path(image_path).with_name(
        Path(image_path).stem + "_sketch.png"
    )
)

cv2.imwrite(out_path, sketch)

plt.figure(figsize=(8, 6))
plt.imshow(sketch, cmap="gray")
plt.axis("off")
plt.show()

print("Sketch saved to:", out_path)
```
