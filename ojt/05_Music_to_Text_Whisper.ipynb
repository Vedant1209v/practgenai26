# Music / Audio to Text (Whisper)

Transcribes speech in an audio file. Music-only audio may not produce meaningful words. Supported formats depend on the installed audio backend.

```python
from pathlib import Path
import torch
from transformers import pipeline

asr = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-small",
    device=0 if torch.cuda.is_available() else -1
)
print("Whisper loaded.")
```

```python
audio_path = input("Enter audio path (e.g. recording.mp3): ").strip().strip('"')
if not Path(audio_path).is_file():
    raise FileNotFoundError(f"Audio file not found: {audio_path}")
result = asr(audio_path, generate_kwargs={"task": "transcribe", "language": "english"})
print("Generated text:", result["text"])
```
