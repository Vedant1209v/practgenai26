# Text to Speech

Uses Google Text-to-Speech (gTTS), which requires an internet connection because synthesis is performed through Google's service.

```python
%pip install -q gTTS
```

```python
from gtts import gTTS
from IPython.display import Audio, display
from pathlib import Path

text = input("Enter text to speak: ").strip()
if not text:
    raise ValueError("Please enter some text.")
output_path = "speech.mp3"
gTTS(text=text, lang="en").save(output_path)
print("Speech saved to:", Path(output_path).resolve())
display(Audio(output_path))
```
