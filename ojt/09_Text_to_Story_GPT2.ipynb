# Text to Story Generation (GPT-2)

A basic text continuation demo. GPT-2 may produce repetitive or inaccurate text; treat the result as creative output, not factual information.

```python
%pip install -q transformers torch
```

```python
import torch
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2", device=0 if torch.cuda.is_available() else -1)
print("GPT-2 loaded.")
```

```python
prompt = input("Enter a story prompt: ").strip()
if not prompt:
    raise ValueError("Please enter a prompt.")
result = generator(prompt, max_new_tokens=100, num_return_sequences=1, do_sample=True, temperature=0.8, top_p=0.95, pad_token_id=50256)
story = result[0]["generated_text"]
print("\nGenerated Story:\n", story)
```
